```python
df['restaurant_name']  
```

```python
df1=df.groupby("restaurant_name").sum(numeric_only=True)
```

```python
df1
```

```python
d2 = df['restaurant_name'].value_counts().head(10)
sns.barplot(x=d2.index, y=d2.values)
plt.title('Top 10 Restaurants by Number of Orders')
plt.xlabel('Restaurant Name')
plt.ylabel('Number of Orders')
plt.xticks(rotation=90)
plt.show()
```

```python
df['cuisine_type']
```

```python
d1=df.groupby("cuisine_type").sum(numeric_only=True)
```

```python
d1
```

```python
sns.lineplot(x="cuisine_type" , y = "cost_of_the_order" , data=d1)
plt.title('cuisine type by cost of Orders')
plt.xlabel('cuisine_type')
plt.ylabel('cost_of_the_order')
plt.xticks(rotation=90)
plt.show()
```

```python
sns.histplot(x="cuisine_type" , y = "delivery_time" , data=d1)
plt.title('cuisine type by delivery time')
plt.xlabel('cuisine_type')
plt.ylabel('delivery_time')
plt.xticks(rotation=60)
plt.show()
```
